```python
#1. Create a Spark Session
```

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("BigData").getOrCreate()
```

```python
#2.Create a DataFrame
```

```python
data = [
    (1,"Sree",22),
    (2,"Bharathi",20),
    (3,"Bhavi",21),
]

columns=["ID","Name","Age"]

df=spark.createDataFrame(data, columns)

df.show()
```

```python
#3.Select a column
```

```python
df.select("Name").show()
```

```python
#4.Filter Data
```

```python
df.filter(df.Age>21).show()
```

```python
#5.Add a New Column
```

```python
from pyspark.sql.functions import lit

df=df.withColumn("country",lit("India"))
df.show()
```

```python
#6.Count Rows
```

```python
print(df.count())
```

```python
#7.Display Schema
```

```python
df.printSchema()
```

```python
#8. Read a CSV File
```

```python
df=spark.read.csv("students.csv",header=True,inferSchema=True)
df.show()
```

```python
#9. Write Data to a CSV File
```

```python
df.write.csv("output",header=True)
```

```python
#10. Stop Spark
```

```python
spark.stop()
```

```python
#Show the DataFrame
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Example1").getOrCreate()

data = [
    (1, "Sreenivas", 45),
    (2, "Anjali", 23),
    (3, "Sandeep", 23),
    (4, "Siva", 27)
]

columns = ["ID", "Name", "Age"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
#2. Display Only One Column
```

```python
df.select("Name").show()
```

```python
#3. Display Multiple Columns
```

```python
df.select("Name", "Age").show()
```

```python
#4. Filter Students Older Than 21
```

```python
df.filter(df.Age > 21).show()
```

```python
#5. Filter by Name
```

```python
df.filter(df.Name == "Sreenivas").show()
```

```python
#6. Count Number of Rows
```

```python
print(df.count())
```

```python
df.printSchema()
```

```python
#8. Show First 2 Rows
```

```python
df.show(2)
```

```python
#9. Add a New Column
```

```python
from pyspark.sql.functions import lit

df=df.withColumn("Country", lit("India"))

df.show()
```

```python
#10. Sort by Age
```

```python
df.orderBy("Age").show()
```

```python
#11.Show Highest Age First
```

```python
df.orderBy(df.Age.desc()).show()
```

```python
#12.Remove Duplicate Rows
```

```python
df.dropDuplicates().show()
```

```python
#13. Rename a Column
```

```python
df=df.withColumnRenamed("Age","Student_Age")

df.show()
```

```python
#14. Drop a Column
```

```python
df=df.drop("Country")

df.show()
```

```python
#15. Display Column Names
```

```python
print(df.columns)
```
